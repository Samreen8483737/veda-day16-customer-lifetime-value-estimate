```python
import pandas as pd
import numpy as np

# 1. Create dummy dataset mimicking 'Online Retail II'
np.random.seed(42)
data = {
    'CustomerID': np.random.randint(1001, 1006, 50), # 5 distinct customers
    'TotalSales': np.random.uniform(50, 500, 50).round(2)
}
df = pd.DataFrame(data)

# 2. Calculate Base Metrics (AOV and Purchase Frequency)
customer_metrics = df.groupby('CustomerID').agg(
    Total_Revenue=('TotalSales', 'sum'),
    Order_Count=('TotalSales', 'count'),
).reset_index()

customer_metrics['Average_Order_Value'] = (customer_metrics['Total_Revenue'] / customer_metrics['Order_Count']).round(2)

# 3. Method 1: Historical CLV (What they have spent so far)
customer_metrics['Historical_CLV'] = customer_metrics['Total_Revenue']

# 4. Method 2: Predictive CLV (What they will spend in the future)
# ASSUMPTION: The average customer lifespan is 3 years.
assumed_lifespan_years = 3
customer_metrics['Predictive_CLV'] = (customer_metrics['Historical_CLV'] * assumed_lifespan_years).round(2)

# 5. Segment Analysis
customer_metrics['Segment'] = pd.qcut(customer_metrics['Predictive_CLV'], 3, labels=['Low Value', 'Medium Value', 'High Value'])

print("--- CUSTOMER LIFETIME VALUE (CLV) TABLE & SEGMENTS ---")
print(customer_metrics)
```
